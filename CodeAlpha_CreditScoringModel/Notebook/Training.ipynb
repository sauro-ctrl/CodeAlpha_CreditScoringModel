# CREDIT SCORING MODEL

---

**Algorithms Used & Tested:**
- Logistic Regression
- **Decision Tree Classifier**
- Random Forest Classifier
- Gradient Boosting Classifier

---

> **Note:** This Model Was Not Trained On Any Real Data. 

### LOAD DATASET

```python
import pandas as pd
import numpy as np

df = pd.read_csv("D:/intern/Data/credit_data.csv")
df = df.drop_duplicates()
df.head()
```

### SPLITTING

```python
from sklearn.model_selection import train_test_split
X = df.drop("Creditworthy", axis=1)
y = df["Creditworthy"]

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

---
### IMPORTING LIBRARIES
---
#### PIPELINE,IMPUTER,SCALER,ENCODER,COLUMN-TRANSFORMER

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

### SET CATEGORICAL & NUMERICAL PIPELINE

```python
categorical_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="most_frequent")),
    ("encoder",OneHotEncoder(handle_unknown="ignore"))
])

numerical_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="mean")),
    ("scaler",StandardScaler())
])
```

### DEFINING CATEGORICAL & NUMERICAL COLUMNS

```python
cat_cols = X.select_dtypes(include="object").columns
num_cols = X.select_dtypes(include="number").columns
```

### SET COLUMN-TRANSFORMER

```python
preprocessor = ColumnTransformer([
    ("categorical",categorical_pipeline,cat_cols),
    ("numerical",numerical_pipeline,num_cols)
])
```

---
### IMPORTING ALGORITHMS
---
#### Logistic Regression, Decision Tree Classifier, Random Forest Classifier, Gradient Boosting Classifier

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
```

### SET MODELS AND PREPROCESSOR IN PIPELINE

```python
model_log_reg = Pipeline([
    ("preprocessor",preprocessor),
    ("log_reg",LogisticRegression())
])

model_dec_tree = Pipeline([
    ("preprocessor",preprocessor),
    ("dec_tree",DecisionTreeClassifier())
])

model_rand_forest = Pipeline([
    ("preprocessor",preprocessor),
    ("rand_forest",RandomForestClassifier())
])

model_grad_boost = Pipeline([
    ("preprocessor",preprocessor),
    ("grad_boost",GradientBoostingClassifier())
])
```

---
### IMPORT METRICES
--- 
#### Accuracy score,Precision score, Recall score, F1 score

```python
from sklearn.metrics import accuracy_score,precision_score, recall_score, f1_score
```

### IMPORT CROSS-VALIDATOR

```python
from sklearn.model_selection import cross_validate,StratifiedKFold
```

---
### CROSS VALIDATION
---

#### SPLITTING CROSS-VALIDATION
#### CROSS VALIDATING DIFFERENT METRICES FOR EACH ALGORITHM:
- Accuracy score,
- Precision score, 
- Recall score, 
- F1 score

```python
cv = StratifiedKFold(
    n_splits = 5,
    shuffle = True,
    random_state=42
)

scores_log_reg = cross_validate(
    model_log_reg,
    X_train,
    y_train,
    cv=cv,
    scoring=["accuracy", "precision", "recall", "f1"]
)


Accuracy_log_reg = scores_log_reg["test_accuracy"].mean()
Precision_log_reg =scores_log_reg["test_precision"].mean()
Recall_log_reg = scores_log_reg["test_recall"].mean()
F1_log_reg = scores_log_reg["test_f1"].mean()

scores_dec_tree = cross_validate(
    model_dec_tree,
    X_train,
    y_train,
    cv=cv,
    scoring=["accuracy", "precision", "recall", "f1"]
)

Accuracy_dec_tree = scores_dec_tree["test_accuracy"].mean()
Precision_dec_tree =scores_dec_tree["test_precision"].mean()
Recall_dec_tree = scores_dec_tree["test_recall"].mean()
F1_dec_tree = scores_dec_tree["test_f1"].mean()

scores_rand_forest = cross_validate(
    model_rand_forest,
    X_train,
    y_train,
    cv=cv,
    scoring=["accuracy", "precision", "recall", "f1"]
)

Accuracy_rand_forest = scores_rand_forest["test_accuracy"].mean()
Precision_rand_forest =scores_rand_forest["test_precision"].mean()
Recall_rand_forest = scores_rand_forest["test_recall"].mean()
F1_rand_forest = scores_rand_forest["test_f1"].mean()

scores_grad_boost = cross_validate(
    model_grad_boost,
    X_train,
    y_train,
    cv=cv,
    scoring=["accuracy", "precision", "recall", "f1"]
)

Accuracy_grad_boost = scores_grad_boost["test_accuracy"].mean()
Precision_grad_boost =scores_grad_boost["test_precision"].mean()
Recall_grad_boost = scores_grad_boost["test_recall"].mean()
F1_grad_boost = scores_grad_boost["test_f1"].mean()
```

### CREATING A DATAFRAME OF THE RESULTS


```python
cv_df = pd.DataFrame({
    "Model":["Log Reg","Dec Tree","Rand_Forest","Grad Boost"],
    "Accuracy":[Accuracy_log_reg,Accuracy_dec_tree,Accuracy_rand_forest,Accuracy_grad_boost],
    "Precision":[Precision_log_reg,Precision_dec_tree,Precision_rand_forest,Precision_grad_boost],
    "Recall":[Recall_log_reg,Recall_dec_tree,Recall_rand_forest,Recall_grad_boost],
    "F1":[F1_log_reg,F1_dec_tree,F1_rand_forest,F1_grad_boost]
})

cv_df
```

### OVERFITTING TEST FOR THE BEST ALGORITHM

```python
scores_overfit_test = cross_validate(
    model_dec_tree,
    X_train,
    y_train,
    cv=cv,
    scoring="accuracy",
    return_train_score=True
)

print("Train:", scores_overfit_test["train_score"].mean())
print("Validation:", scores_overfit_test["test_score"].mean())
```

```python
print("Got the highest (0.997753) Accuracy with Decision Tree Classifire Algorithm using 5-Folds of Cross Validation")
print("Tested for any overfitting with accuracy validation and train score. Which clearly shows there is no sign of overfitting")
```

### FITTING THE BEST ALGORITHM

```python
model_dec_tree.fit(X_train,y_train)
```

### PREDICT

```python
y_pred_dec_tree = model_dec_tree.predict(X_test)
```

### METRICES

```python
acc_dec_tree = accuracy_score(y_test,y_pred_dec_tree)
prec_dec_tree = precision_score(y_test,y_pred_dec_tree)
rec_dec_tree = recall_score(y_test,y_pred_dec_tree)
f1_dec_tree = f1_score(y_test,y_pred_dec_tree)
```

```python
print("Accuracy:",acc_dec_tree)
print("Precision:",prec_dec_tree)
print("Recall:",rec_dec_tree)
print("F1:",f1_dec_tree)
```
